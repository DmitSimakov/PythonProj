#Приложение «MyProfile для предпринимателей»
Вы получили свой первый заказ на фрилансе, в рамках которого нужно доработать
имеющееся у заказчика консольное приложение. Сейчас в приложении MyProfile
пользователь может сохранять личную информацию о себе (имя, возраст, телефон,
электронная почта, профили социальных сетей и мессенджеров, дополнительная
информация) и выводить её в различных форматах. Но заказчик хочет расширить
продуктовую линейку и выпустить бизнес-версию этого приложения.
В приложении «MyProfile для предпринимателей» будет храниться информация
о предпринимателе (например, ИНН), а также его банковские реквизиты. Такую
информацию удобно предоставлять для заключения договоров и дополнительных
соглашений.
Задачи
1. Удалите ввод, хранение и вывод информации о социальных сетях и
мессенджерах.
2. Добавьте ввод полей ОГРНИП и ИНН. При вводе проверьте, что поле ОГРНИП
содержит 15 цифр.
3. Добавьте ввод банковских реквизитов: расчётный счёт, название банка, БИК и
корреспондентский счёт. При вводе банковских реквизитов проверьте, что
расчётный счёт содержит 20 цифр.
4. Добавьте в вывод полной информации поля ОГРНИП, ИНН, а также
информацию о банковских реквизитах.
5. Добавьте ввод и вывод почтового адреса. Индекс хранится в отдельной
переменной. Индекс вводится как произвольная строка, но в приложении
сохраняются только цифры.


```python
# -*- coding: utf-8 -*-

def input_digit_field(prompt, length=None, allow_negative=False):
    """
    Ввод числового поля с проверкой.
    :param prompt: Подсказка для ввода
    :param length: Требуемая длина числа (количество цифр)
    :param allow_negative: Разрешить отрицательные числа
    :return: Введенное число (int)
    """
    while True:
        try:
            value = int(input(prompt))
            if not allow_negative and value < 0:
                print("Возраст не может быть отрицательным. Попробуйте снова.")
                continue
            if length is not None and len(str(value)) != length:
                print(f"Должно содержать ровно {length} цифр. Попробуйте снова.")
                continue
            return value
        except ValueError:
            print("Некорректный ввод. Введите целое число.")

def input_string_field(prompt, allow_empty=False):
    """
    Ввод строкового поля.
    :param prompt: Подсказка для ввода
    :param allow_empty: Разрешить пустую строку
    :return: Введенная строка
    """
    while True:
        value = input(prompt).strip()
        if not allow_empty and not value:
            print("Поле не может быть пустым. Попробуйте снова.")
            continue
        return value

def input_postal_index():
    """
    Ввод почтового индекса с сохранением только цифр.
    :return: Индекс, содержащий только цифры
    """
    while True:
        index = input("Индекс: ").strip()
        digits_only = ''.join(filter(str.isdigit, index))
        if digits_only:
            return digits_only
        else:
            print("Индекс должен содержать хотя бы одну цифру. Попробуйте снова.")

def input_personal_info():
    """
    Ввод или обновление личной информации.
    """
    print("\nВведите личную информацию:")
    personal_info["name"] = input_string_field("Имя: ")
    personal_info["age"] = input_digit_field("Возраст: ", allow_negative=False)
    personal_info["phone"] = input_string_field("Телефон: ")
    personal_info["email"] = input_string_field("Электронная почта: ")
    personal_info["postal_index"] = input_postal_index()
    personal_info["postal_address"] = input_string_field("Почтовый адрес: ", allow_empty=True)
    personal_info["additional_info"] = input_string_field("Дополнительная информация: ", allow_empty=True)
    print("Личная информация сохранена.\n")

def input_business_info():
    """
    Ввод или обновление информации о предпринимателе.
    """
    print("\nВведите информацию о предпринимателе:")
    business_info["ogrnip"] = input_digit_field("ОГРНИП (15 цифр): ", length=15)
    business_info["inn"] = input_digit_field("ИНН: ")
    business_info["account"] = input_digit_field("Расчётный счёт (20 цифр): ", length=20)
    business_info["bank_name"] = input_string_field("Название банка: ")
    business_info["bik"] = input_digit_field("БИК: ")
    business_info["correspondent_account"] = input_digit_field("Корреспондентский счёт: ")
    print("Информация о предпринимателе сохранена.\n")

def print_personal_info():
    """
    Вывод личной информации.
    """
    if not personal_info.get("name"):
        print("Личная информация не введена.\n")
        return
    print("\nЛичная информация:")
    print(f"Имя: {personal_info['name']}")
    print(f"Возраст: {personal_info['age']}")
    print(f"Телефон: {personal_info['phone']}")
    print(f"Электронная почта: {personal_info['email']}")
    print(f"Индекс: {personal_info['postal_index']}")
    print(f"Почтовый адрес: {personal_info['postal_address']}")
    print(f"Дополнительная информация: {personal_info['additional_info']}\n")

def print_full_info():
    """
    Вывод всей информации (личной и бизнес).
    """
    print_personal_info()
    if not business_info.get("inn"):
        print("Информация о предпринимателе не введена.\n")
        return
    print("Информация о предпринимателе:")
    print(f"ОГРНИП: {business_info['ogrnip']}")
    print(f"ИНН: {business_info['inn']}")
    print(f"Расчётный счёт: {business_info['account']}")
    print(f"Название банка: {business_info['bank_name']}")
    print(f"БИК: {business_info['bik']}")
    print(f"Корреспондентский счёт: {business_info['correspondent_account']}\n")

def main_menu():
    """
    Главное меню приложения.
    """
    while True:
        print("Главное меню:")
        print("1 — Ввести или обновить информацию")
        print("2 — Вывести информацию")
        print("0 — Завершить работу")
        try:
            choice = int(input("Введите номер пункта меню: "))
        except ValueError:
            print("Введён некорректный пункт меню\n")
            continue

        if choice == 1:
            input_menu()
        elif choice == 2:
            output_menu()
        elif choice == 0:
            print("Работа приложения завершена.")
            break
        else:
            print("Введён некорректный пункт меню\n")

def input_menu():
    """
    Меню ввода информации.
    """
    while True:
        print("\nВвод или обновление информации:")
        print("1 — Личная информация")
        print("2 — Информация о предпринимателе")
        print("0 — Назад")
        try:
            choice = int(input("Введите номер пункта меню: "))
        except ValueError:
            print("Введён некорректный пункт меню")
            continue

        if choice == 1:
            input_personal_info()
        elif choice == 2:
            input_business_info()
        elif choice == 0:
            break
        else:
            print("Введён некорректный пункт меню")

def output_menu():
    """
    Меню вывода информации.
    """
    while True:
        print("\nВывод информации:")
        print("1 — Личная информация")
        print("2 — Вся информация")
        print("0 — Назад")
        try:
            choice = int(input("Введите номер пункта меню: "))
        except ValueError:
            print("Введён некорректный пункт меню")
            continue

        if choice == 1:
            print_personal_info()
        elif choice == 2:
            print_full_info()
        elif choice == 0:
            break
        else:
            print("Введён некорректный пункт меню")

if __name__ == "__main__":
    # Словари для хранения данных
    personal_info = {}
    business_info = {}
    print("Добро пожаловать в приложение «MyProfile для предпринимателей»!\n")
    main_menu()
```
